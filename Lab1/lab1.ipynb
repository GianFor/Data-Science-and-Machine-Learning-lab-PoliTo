### Feature standardization
This step consists of transforming the Iris feature matrix so thath each col (feature) is standardized to have zero mean and unit variance. **This ensures
that all measurements contribute equally to subsequent analyses, regardless of
their original scale.**

**Computing the standardized matrix**
Use NumPy’s aggregation func-
tions to compute the column-wise mean and standard deviation of the
feature matrix X. Then, apply the standardization formula:

$$Z = \frac{X - mean(X)}{std(X)}$$

to create a new matrix Z of the same shape as X. Confirm that the trans-
formation has been applied independently to each feature column.

````python
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)

print(X_mean.shape, X_std.shape)
````

````python
Z = (X - X_mean)/X_std
print(np.mean(Z))
print(np.std(Z))
````

### Naive single-feature classifier
In this final exercise on the Iris dataset, you will build a simple classifier that
uses only one feature to predict the species of a flower.

````python
classes = np.unique(y)
means = np.vstack([X[y==c].mean(axis=0) for c in classes])
print(means)
````

````python
from itertools import combinations

class_pairs = np.array(list(combinations(range(len(classes)), 2)))

pair_diffs = np.abs(means[class_pairs[:,0], :] - means[class_pairs[:,1], :])

min_sep_by_feat = np.min(pair_diffs, axis=0)
print(min_sep_by_feat)

best_feat = int(np.argmax(min_sep_by_feat))
best_feat_name = feature_names[best_feat]
print(f"Best feature: {best_feat} ({best_feat_name})")
````

````python
m = means[:,best_feat]
order = np.argsort(m)
m_sorted = m[order]
classes_sorted = classes[order]

print(m_sorted)
print(classes_sorted)
````

````python
t1 = 0.5*(m_sorted[0] + m_sorted[1])
t2 = 0.5*(m_sorted[1] + m_sorted[2])

print(t1,t2)
````

````python
feature_values = X[:, best_feat]
predictions = np.where(feature_values < t1, 0, np.where(feature_values < t2, 1, 2))

prediction_classes = classes_sorted[predictions]

acc = (prediction_classes == y).mean()
print(acc)
````
